数组的拼接

```python
grid = np.array([[1, 2, 3],
                [4, 5, 6]])
print(np.concatenate([grid, grid]))    #沿第一个轴拼接
print(np.concatenate([grid, grid], axis=1))   #沿第二个轴拼接
```

沿着固定维度处理数组时，使用 np.vstack（垂直栈）和 np.hstack（水平栈）函数会更简洁：

```python
x = np.array([1, 2, 3])
grid = np.array([[9, 8, 7],
                 [6, 5, 4]])
print(np.vstack([x, grid]))
y = np.array([[99],
              [99]])
print(np.hstack([grid, y]))
```

拆分数组

```python
x = [1, 2, 3, 99, 99, 3, 2, 1]
x1, x2, x3 = np.split(x, [3, 5])
print(x1, x2, x3)
```

相关的 np.hsplit和 np.vsplit的用法也类似：

```python
grid = np.arange(16).reshape((4, 4))
print(grid)
upper, lower = np.vsplit(grid, [2])
print(upper)
print(lower)
left, right = np.hsplit(grid, [2])
print(left)
print(right)
```

python作为解释型语言，执行效率偏低

```python
import numpy as np
np.random.seed(0)
def compute_reciprocals(values):
    output = np.empty(len(values))
    for i in range(len(values)):
        output[i] = 1.0 / values[i]
    return output
values = np.random.randint(1, 10, size=5)
print(compute_reciprocals(values))

# big_array = np.random.randint(1, 100, size=1000000)
# %timeit compute_reciprocals(big_array)
```

通用函数运算

1.数值运算

```python
x = np.arange(4)
print("x     =", x)   
print("x + 5 =", x + 5)    #np.add
print("x - 5 =", x - 5)    #np.substract
print("x * 2 =", x * 2)    #np.multiply
print("x / 2 =", x / 2)    #np.divide
print("x // 2 =", x // 2)  #np.flood_divide 
print("x ** 2 =", x ** 2)  #np.power
```

2.绝对值 

```python
x = np.array([-2, -1, 0, 1, 2])
print(np.abs(x))
y = np.array([3 - 4j, 4 - 3j, 2 + 0j, 0 + 1j])
print(np.abs(y))     # 对复数取模
```

3.三角函数/反三角函数

```python
theta = np.linspace(0, np.pi, 3)
print("theta      = ", theta)
print("sin(theta) = ", np.sin(theta))
print("cos(theta) = ", np.cos(theta))
print("tan(theta) = ", np.tan(theta))
```

这些值是在机器精度内计算的，所以有些应该是 0 的值并没有精确到 0

```python
x = [-1, 0, 1]
print("x         = ", x)
print("arcsin(x) = ", np.arcsin(x))
print("arccos(x) = ", np.arccos(x))
print("arctan(x) = ", np.arctan(x))
```

其他函数

```python
from scipy import special
```

```python
# 误差函数（高斯积分）
x = np.array([0, 0.3, 0.7, 1.0])
print("erf(x)  =", special.erf(x))
print("erfc(x) =", special.erfc(x))
print("erfinv(x) =", special.erfinv(x))
```

```python
# Gamma函数（广义阶乘，generalized factorials）和相关函数
x = [1, 5, 10]
print("gamma(x)     =", special.gamma(x))
print("ln|gamma(x)| =", special.gammaln(x))
print("beta(x, 2)   =", special.beta(x, 2))
```

高级通用函数

1.指定输出

```python
#不同于创建临时数组，你可以用这个特性将计算结果直接写入到你期望的存储位置。
#所有的通用函数都可以通过 out参数来指定计算结果的存放位置不同于创建临时数组，你可以用这个特性将计算结果直接写入到你期望的存储位置。
x = np.arange(5)
y = np.empty(5)
np.multiply(x, 10, out=y)
print(y)
```

2.聚合

```python
#一个 reduce方法会对给定的元素和操作重复执行，直至得到单个的结果。
x = np.arange(1,6)
print(np.add.reduce(x))
print(np.multiply.reduce(x))
```

```python
#accumulate可用于储存中间结果
print(np.add.accumulate(x))
```

3.外积

```python
#任何通用函数都可以用 outer方法获得两个不同输入数组所有元素对的函数运算结果
print(np.multiply.outer(x, x))
```

基本统计函数：求和，平均数，中位数，方差，标准差，Max,Min，分位数，

```python
np.random.seed(0)
x = np.random.randint(1,100,8)
print(x)
print(np.sum(x))
print(np.mean(x))  #平均数
print(np.median(x))  #中位数
print(np.var(x))  #方差
print(np.std(x))    #标准差
print(np.max(x),np.argmax(x))   #查找最大值，最大值位置
print(np.min(x),np.argmin(x))
print(np.percentile(x,20))
```
