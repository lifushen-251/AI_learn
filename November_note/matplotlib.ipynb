```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'
```

```python
#创建图表并设置大小
plt.figure(figsize = (10,5))

#图表数据
month = ['1月', '2月', '3月', '4月']
sales = [100, 40 ,120, 60]

#绘制图表
plt.plot(month,sales
         ,label = '产品A')

#添加标题
plt.title('114514',color = 'red', fontsize = 20)

#添加坐标轴标签
plt.xlabel('month')
plt.ylabel('销售额')

#添加图例
plt.legend(loc = 'upper left')
plt.grid(True, alpha = 0.1, color = 'blue', linestyle = '--')  #plt.grid(axis = 'x')

#显示图表
plt.show()
```
